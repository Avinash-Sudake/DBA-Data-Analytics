```python
# Program For Gread System

marks=int(input("Enter a Marks = "))
if 91<=marks<=100:
  print("o Grade")
elif 81<=marks<=90:
  print("A Grade")
elif 71<=marks<=80:
  print("B Grade")
elif 61<=marks<=70:
  print("C Grade")
elif marks<60:
  print("Fail")
else:
  print(" Invalid Input ")
```
