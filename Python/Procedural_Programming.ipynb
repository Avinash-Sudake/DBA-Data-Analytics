```python
# create a menu system taking
# multiple orders and printing
# total bill after the order is
# completed

print('-------MENU------')
print('1. Vadapav')
print('2. Samosa')
print('3. Tea')
print('4. Poha')
cont = 'y'
total = []
q = {}
price = {'vadapav': 15,'samosa':20,'tea': 12, 'poha' : 25 }

while True:

  order = input('Enter The Order Number (Exit:[n]) : ')

  if order == 'n':
    break

  if int(order) == 1:  # false
    qty = int(input('Enter The Quantity : '))
    total.append(15*qty)
    q['vadapav'] = qty
  elif int(order) == 2:  # False
    qty = int(input('Enter The Quantity : '))
    total.append(20*qty)
    q['samosa'] = qty
  elif int(order) == 3:  # False
    qty = int(input('Enter The Quantity : '))
    total.append(12*qty)
    q['tea'] = qty
  elif int(order) == 4:   # False
    qty = int(input('Enter The Quantity : '))
    total.append(25*qty)
    q['poha'] = qty
  else:
    print('Order does not exist')
    total.append(0)

print('---- Total Bill----')
for i,j in q.items():
  print(f'{i} x {j} = {price[i] * j}')
print('The Total Bill is  = ', sum(total))
```

```python
num = 1635
z = num
l = len(str(num))   # digits 3
arm = 0
while num > 0:
  n = num%10     # n = 1
  arm = arm + n**l  # 153
  num = num//10  # num//10 0

# print(num)
if arm == z:
  print('Armstrong Number')
else:
  print('Not a Armstrong Number')
```

```python
def revv(n):
  rev = 0
  while n > 0:
    num = n % 10 #  6, 7, 8
    rev = rev *10 + num   # 6,67,678
    n = n//10   # 87, 8, 0

  return rev

number = int(input('Enter The number that needs to be reversed :'))
print(revv(number))

```

```python
# Arbitary Argument Function
def add(*n):
  sum = 0
  for i in n:
    sum = sum + i

  return sum
print(add(5,1,2,3,4,5,6,7))
```

```python
def student_info(**d):
  for i,j in d.items():
    print(f'{i} - {j}')

student_info(Rollno = 1,Name ='Runal',Class = 'FYMCA')
```

```python
# Lambda Funcions
# lambda - also called as anonymous function
#        - it is defined using lambda
#        - it is an one line function
#        - lambda function by default run
#          return
```

```python
x = lambda a : a**2
print(x(7))
```

```python
y = lambda a,b : a+b
print((lambda a,b : a+b)(8,2))
```

```python
v = [2,4,6]
t = lambda x : [i**2 for i in x]
print(t(v))
```

```python
# Write a Program to find the length
# of each element in a list with the help
# of lambda function
b = ['hello','world','python']
m = lambda x: [len(i) for i in x]
print(m(b))
```

```python
# Write a Progam to find even and odd
# numbers from a list with help of
# lambda function

w = [7,4,9,1,10,13,16,17]
h = lambda x: [f'Even:{i}' if i % 2 == 0 else
               f'Odd:{i}' for i in x]
print(h(w))
```
