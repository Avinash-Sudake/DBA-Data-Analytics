```python
# Indexing
z = "Python programming"
print("Find a = ",z[12])
print("Find Space = ",z[6])
print("Find n = ",z[-13])
print("Find h = ",z[-15])
```

```python
# Slicing (Positive) [min:max +1]
z = "Python programming"
print(z[0:14])
print(z[0:])
print(z[:18])
print(z[:])
```

```python
# Slicing (Negative) [min:max ]
z = "Python programming"
print(z[-11:])
```

```python
# Stepping
z = "Python programming"
print(z[3:14:2])
print(z[3:14:3])
```
